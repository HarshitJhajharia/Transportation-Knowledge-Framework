```python
import networkx as nx
```

```python
G = nx.Graph()

roads = [
    ("A", "B", 4),
    ("A", "C", 3),
    ("B", "D", 5),
    ("B", "E", 2),
    ("C", "E", 4),
    ("C", "G", 6),
    ("D", "F", 3),
    ("F", "G", 2)
]

G.add_weighted_edges_from(roads)

print("Transportation network created!")
```

```python
source = "A"
destination = "F"

original_path = nx.shortest_path(G, source, destination, weight="weight")

original_distance = nx.shortest_path_length(
    G, source, destination, weight="weight"
)

print("Original Route:", " -> ".join(original_path))
print("Original Distance:", original_distance, "km")
```

```python
incident = {
    "type": "Accident",
    "road": ("C", "G"),
    "status": "Blocked"
}

if incident["status"] == "Blocked":
    G.remove_edge(*incident["road"])
    print("Incident detected!")
    print("Road C-G is now blocked.")
```

```python
new_path = nx.shortest_path(G, source, destination, weight="weight")

new_distance = nx.shortest_path_length(
    G, source, destination, weight="weight"
)

extra_distance = new_distance - original_distance

print("Alternative Route:", " -> ".join(new_path))
print("Alternative Distance:", new_distance, "km")
print("Additional Distance:", extra_distance, "km")
```

```python
def explain_route(incident, original_path, new_path, original_distance, new_distance):

    road = incident["road"]

    print("TRANSPORTATION DECISION EXPLANATION")
    print("-----------------------------------")

    print("Incident:", incident["type"])
    print("Affected Road:", road[0] + "-" + road[1])
    print("Road Status:", incident["status"])

    print("\nOriginal Route:", " -> ".join(original_path))
    print("Original Distance:", original_distance, "km")

    print("\nRecommended Route:", " -> ".join(new_path))
    print("New Distance:", new_distance, "km")

    print("\nREASON:")
    print(
        f"An {incident['type'].lower()} has blocked road "
        f"{road[0]}-{road[1]}. "
        "The original shortest route used this road, "
        "so an alternative route was calculated."
    )

    print(
        f"The new route is {new_distance - original_distance} km longer "
        "but avoids the blocked road."
    )


explain_route(
    incident,
    original_path,
    new_path,
    original_distance,
    new_distance
)
```

```python
print("CAUSAL EXPLANATION")
print("------------------")

print("1. An accident occurred on road C-G.")
print("2. The accident caused road C-G to become blocked.")
print("3. The blockage may contribute to congestion on nearby roads.")
print("4. The original shortest route is no longer available.")
print("5. The system recommends A -> B -> D -> F instead.")
```
