```python
import pandas as pd
import xml.etree.ElementTree as ET
from pathlib import Path
```

```python
project_folder = Path.cwd()

if project_folder.name == "notebooks":
    project_folder = project_folder.parent

file = project_folder / "data" / "outputs" / "traffic_edgedata.xml"

print("File exists:", file.exists())
```

```python
tree = ET.parse(file)
root = tree.getroot()

data = []

for interval in root.findall("interval"):
    for edge in interval.findall("edge"):
        data.append({
            "begin": float(interval.get("begin")),
            "end": float(interval.get("end")),
            "roadId": edge.get("id"),
            "avgSpeed": float(edge.get("speed", 0)) * 3.6,
            "density": float(edge.get("density", 0)),
            "vehicleCount": int(edge.get("entered", 0))
        })

df = pd.DataFrame(data)

print("Traffic records:", len(df))
df.head()
```

```python
output = project_folder / "data" / "outputs" / "traffic_data.csv"

df.to_csv(output, index=False)

print("Traffic data saved successfully!")
print("Records:", len(df))
```
