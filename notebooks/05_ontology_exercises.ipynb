```python
from owlready2 import *
from pathlib import Path

project_folder = Path.cwd()

if project_folder.name == "notebooks":
    project_folder = project_folder.parent

ontology_path = project_folder / "ontology" / "transportation.owl"

onto = get_ontology("http://example.org/transportation.owl")

onto.load(fileobj=open(ontology_path, "rb"))

print("Ontology loaded successfully!")
```

```python
with onto:

    class Car(onto.Vehicle):
        pass

    class EmergencyVehicle(onto.Vehicle):
        pass

    class Ambulance(EmergencyVehicle):
        pass

print("Vehicle subclasses created!")
```

```python
car = onto.Car("Car_1")
emergency_vehicle = onto.EmergencyVehicle("EmergencyVehicle_1")
ambulance = onto.Ambulance("Ambulance_2")

print("Vehicle instances created!")
```

```python
print("Is Ambulance an EmergencyVehicle?",
      issubclass(onto.Ambulance, onto.EmergencyVehicle))

print("Is Ambulance a Vehicle?",
      issubclass(onto.Ambulance, onto.Vehicle))

print("Is Car an EmergencyVehicle?",
      issubclass(onto.Car, onto.EmergencyVehicle))
```

```python
print("Subclasses of Vehicle:")

for cls in onto.Vehicle.subclasses():
    print(cls.name)

print("\nSubclasses of EmergencyVehicle:")

for cls in onto.EmergencyVehicle.subclasses():
    print(cls.name)
```

```python
output = project_folder / "ontology" / "transportation_extended.owl"

onto.save(file=str(output), format="rdfxml")

print("Extended ontology saved successfully!")
```

```python
with onto:

    class hasRole(DataProperty, FunctionalProperty):
        domain = [onto.Vehicle]
        range = [str]

print("Vehicle role property created!")
```

```python
with onto:

    onto.EmergencyVehicle.equivalent_to.append(
        onto.Vehicle & onto.hasRole.value("Emergency")
    )

print("Emergency vehicle classification rule created!")
```

```python
vehicle = onto.Vehicle("Vehicle_100")

vehicle.hasRole = "Emergency"

print("Vehicle created!")
print("Vehicle role:", vehicle.hasRole)
print("Current types:", vehicle.is_a)
```

```python
print("Running ontology reasoner...")

with onto:
    sync_reasoner()

print("Reasoning completed!")
```

```python
print("Vehicle types after reasoning:")

for cls in vehicle.is_a:
    print(cls.name)

print("\nIs the vehicle an EmergencyVehicle?",
      onto.EmergencyVehicle in vehicle.is_a)
```

```python
output = project_folder / "ontology" / "transportation_reasoned.owl"

onto.save(file=str(output), format="rdfxml")

print("Reasoned ontology saved successfully!")
```
