```python
with onto:

    class ConnectedTo(ObjectProperty):
        domain = [onto.Road]
        range = [onto.Junction]

    class Causes(ObjectProperty):
        domain = [onto.Incident]
        range = [onto.Congestion]

    class LocatedAt(ObjectProperty):
        domain = [onto.Vehicle | onto.Incident | onto.Signal]
        range = [onto.Road | onto.Junction]

    class Influences(ObjectProperty):
        domain = [onto.Congestion]
        range = [onto.Route]

    class Blocks(ObjectProperty):
        domain = [onto.Incident]
        range = [onto.Road]

    class Prioritizes(ObjectProperty):
        domain = [onto.Signal]
        range = [onto.EmergencyVehicle]

print("All six object properties created successfully!")
```

```python
with onto:

    class roadName(DataProperty, FunctionalProperty):
        domain = [onto.Road]
        range = [str]

    class length(DataProperty, FunctionalProperty):
        domain = [onto.Road]
        range = [float]

    class speedLimit(DataProperty, FunctionalProperty):
        domain = [onto.Road]
        range = [int]

    class lanes(DataProperty, FunctionalProperty):
        domain = [onto.Road]
        range = [int]

print("Road data properties created!")
```

```python
with onto:

    class junctionId(DataProperty, FunctionalProperty):
        domain = [onto.Junction]
        range = [str]

    class vehicleId(DataProperty, FunctionalProperty):
        domain = [onto.Vehicle]
        range = [str]

    class signalState(DataProperty, FunctionalProperty):
        domain = [onto.Signal]
        range = [str]

    class signalPhase(DataProperty, FunctionalProperty):
        domain = [onto.Signal]
        range = [str]

    class routeId(DataProperty, FunctionalProperty):
        domain = [onto.Route]
        range = [str]

    class routeDistance(DataProperty, FunctionalProperty):
        domain = [onto.Route]
        range = [float]

    class congestionLevel(DataProperty, FunctionalProperty):
        domain = [onto.Congestion]
        range = [str]

    class avgSpeed(DataProperty, FunctionalProperty):
        domain = [onto.Congestion]
        range = [float]

    class incidentType(DataProperty, FunctionalProperty):
        domain = [onto.Incident]
        range = [str]

    class severity(DataProperty, FunctionalProperty):
        domain = [onto.Incident]
        range = [int]

    class startTime(DataProperty, FunctionalProperty):
        domain = [onto.Incident]
        range = [str]

print("Remaining data properties created!")
```

```python
with onto:

    onto.Road.is_a.append(
        onto.ConnectedTo.min(2, onto.Junction)
    )

print("Road restriction added!")
```

```python
with onto:

    AllDisjoint([
        onto.Road,
        onto.Junction,
        onto.Vehicle,
        onto.Signal,
        onto.Route,
        onto.Congestion,
        onto.Incident
    ])

print("Disjointness axioms added!")
```

```python
junction_a = onto.Junction_A
junction_b = onto.Junction_B
junction_c = onto.Junction_C
junction_g = onto.Junction_G

road_ab = onto.Road_A_B
road_cg = onto.Road_C_G

accident = onto.Accident_1
ambulance = onto.Ambulance_1
signal = onto.Signal_B

print("Existing transportation individuals loaded!")
```

```python
road_ab.roadName = "Road A-B"
road_ab.length = 4.0
road_ab.speedLimit = 60
road_ab.lanes = 2

road_cg.roadName = "Road C-G"
road_cg.length = 6.0
road_cg.speedLimit = 50
road_cg.lanes = 2

print("Road properties updated!")
```

```python
road_ab.ConnectedTo = [junction_a, junction_b]

road_cg.ConnectedTo = [junction_c, junction_g]

print("Road-junction relationships created!")
```

```python
roads = [
    ("A", "B", 4),
    ("A", "C", 3),
    ("B", "D", 5),
    ("B", "E", 2),
    ("C", "E", 4),
    ("C", "G", 6),
    ("D", "F", 3),
    ("F", "G", 2)
]

for a, b, distance in roads:

    road = onto["Road_" + a + "_" + b]

    road.ConnectedTo = [
        onto["Junction_" + a],
        onto["Junction_" + b]
    ]

    road.roadName = f"Road {a}-{b}"
    road.length = float(distance)

    if road.speedLimit is None:
        road.speedLimit = 60

    if road.lanes is None:
        road.lanes = 2

print("All eight roads updated successfully!")
```

```python
congestion = onto.Congestion("Congestion_1")

congestion.congestionLevel = "High"
congestion.avgSpeed = 15.0

route = onto.Route("Route_1")

route.routeId = "R001"
route.routeDistance = 12.0

print("Congestion and Route individuals created!")
```

```python
accident.Blocks = [road_cg]

accident.Causes = [congestion]

congestion.Influences = [route]

print("Incident relationships created!")
```

```python
ambulance.is_a.append(onto.Ambulance)

ambulance.LocatedAt = [junction_a]

signal.LocatedAt = [junction_b]

signal.Prioritizes = [ambulance]

print("Emergency transportation relationships created!")
```

```python
for name in ["A", "B", "C", "D", "E", "F", "G"]:
    junction = onto["Junction_" + name]
    junction.junctionId = name

ambulance.vehicleId = "AMB001"

signal.signalState = "Green"
signal.signalPhase = "Emergency Priority"

accident.incidentType = "Accident"
accident.severity = 3
accident.startTime = "2026-10-27T10:30:00"

print("Additional transportation attributes added!")
```

```python
print("EMERGENCY VEHICLE CLASSIFICATION")
print("--------------------------------")

ambulance = onto.Ambulance_1

print("Individual:", ambulance.name)
print("Classes:", ambulance.is_a)

print(
    "Is an EmergencyVehicle:",
    isinstance(ambulance, onto.EmergencyVehicle)
)

print(
    "Is a Vehicle:",
    isinstance(ambulance, onto.Vehicle)
)
```

```python
print("ROAD-JUNCTION VALIDATION")
print("------------------------")

all_valid = True

for road in onto.Road.instances():

    junctions = road.ConnectedTo

    if len(set(junctions)) >= 2:
        print(road.name, "- VALID")
    else:
        print(road.name, "- INVALID")
        all_valid = False

print("\nAll roads valid:", all_valid)
```
